# 01 - Two-Asset Intuition

The smallest non-trivial portfolio problem has just two risky assets. Everything we do later is a generalization of the geometry on this page.

## Setup

Two assets with known means, volatilities, and correlation.

```python
import matplotlib.pyplot as plt
import numpy as np

mu = np.array([0.06, 0.10])           # annual excess returns
sig = np.array([0.12, 0.20])          # annual volatilities
rho = 0.15                            # correlation
Sigma = np.array([
    [sig[0] ** 2,           rho * sig[0] * sig[1]],
    [rho * sig[0] * sig[1], sig[1] ** 2          ],
])
```

## Sweep weights from 0 to 1 in asset 2

Each $w_2 \in [0, 1]$ defines a fully-invested long-only portfolio $w = (1 - w_2, w_2)$ with mean and standard deviation given by the usual closed forms.

```python
w2 = np.linspace(0.0, 1.0, 201)
w = np.column_stack([1 - w2, w2])
port_mu  = w @ mu
port_var = np.einsum('ij,jk,ik->i', w, Sigma, w)
port_sd  = np.sqrt(port_var)
```

## Plot the parametric frontier

The curve in $(\sigma, \mu)$-space is the **mean-variance frontier**. Its leftmost point is the minimum-variance portfolio.

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(port_sd, port_mu, lw=2)
ax.scatter(sig, mu, c='black', zorder=3)
for i, label in enumerate(['A', 'B']):
    ax.annotate(label, (sig[i], mu[i]), xytext=(6, 6), textcoords='offset points')

min_idx = int(np.argmin(port_sd))
ax.scatter(port_sd[min_idx], port_mu[min_idx], c='crimson', zorder=4, label='min-var')
ax.set_xlabel('portfolio standard deviation')
ax.set_ylabel('portfolio expected return')
ax.set_title('Two-asset mean-variance frontier')
ax.legend()
fig.tight_layout()
```

## Verify the closed form for minimum variance

For two assets,
$$w_2^{\min\text{-}\mathrm{var}} = \frac{\sigma_1^2 - \rho \sigma_1 \sigma_2}{\sigma_1^2 + \sigma_2^2 - 2 \rho \sigma_1 \sigma_2}.$$

```python
w2_star = (sig[0] ** 2 - rho * sig[0] * sig[1]) / (
    sig[0] ** 2 + sig[1] ** 2 - 2 * rho * sig[0] * sig[1]
)
print(f'closed-form w2*  = {w2_star:.4f}')
print(f'grid argmin w2   = {w2[min_idx]:.4f}')
```

## Takeaway

Even at $n = 2$, the optimum depends on the correlation $\rho$ in a non-monotonic way. The next notebook generalizes this picture to arbitrary $n$ using the library's `MeanVariance` optimizer.
